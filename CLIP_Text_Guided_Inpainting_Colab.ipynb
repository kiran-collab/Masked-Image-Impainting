# 🧠 Text-Guided Occluded Image Completion using CLIP and Stable Diffusion
This Colab notebook fills in occluded image regions using Stable Diffusion inpainting guided by a text prompt.

```python
from diffusers import StableDiffusionInpaintPipeline
import torch
from PIL import Image
import requests
```

```python
# Load the Stable Diffusion inpainting pipeline
pipe = StableDiffusionInpaintPipeline.from_pretrained(
    "runwayml/stable-diffusion-inpainting", torch_dtype=torch.float16
).to("cuda")
```

```python
# Download a sample image and mask
image_url = "https://raw.githubusercontent.com/CompVis/stable-diffusion/main/assets/stable-samples/img2img/sketch-mountains-input.jpg"
mask_url = "https://raw.githubusercontent.com/CompVis/stable-diffusion/main/assets/stable-samples/img2img/sketch-mountains-mask.png"

init_image = Image.open(requests.get(image_url, stream=True).raw).convert("RGB").resize((512, 512))
mask_image = Image.open(requests.get(mask_url, stream=True).raw).convert("L").resize((512, 512))
```

```python
# Provide a text prompt
prompt = "a man wearing a blue jacket standing on a snowy mountain"
```

```python
# Run inpainting
result = pipe(prompt=prompt, image=init_image, mask_image=mask_image).images[0]
result.save("output_result.png")
result.show()
```
